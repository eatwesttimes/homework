## Bài 1
### In các số dương lẻ bé hơn 100

```python
a = int(input())
while a % 2 == 0:
    print("Không phải số lẻ")
    a = int(input())
while a % 2 == 1:
    print(a)    
    a = a + 2
    if a >= 100:
        a -= 2
        break
```

## Bài 4
### Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím

```python
h = int(input())
for i in range(1, h+1):
    space = " " * (h - i)
    if i == 1:
        print(space + "*")
    elif i == h:
        print("*"*(2*i-1))
    else:
        space2 = " "*(2*i-3)
        print(space + "*" + space2 + "*")
```

## Bài 12
### Nhập số nguyên dương n. Cho S(k) = 1 + 2 + 3 + ... + k. Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n

```python
n = int(input())
S = 0
for k in range(1, n+1):
    S += k
    if S >= n:
        S -= k
        break
print(f"Giá trị lớn nhất của S(k) thỏa mãn yêu cầu bài toán là:", S)
```

## Bài 13
### Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho: 1 + 1/2 + 1/3 + 1/4 + ... 1/n > A

```python
A = int(input())
S = 0
n = 0
while S <= A:
    n += 1
    S += 1/n
    if S > A:
        print(n)
        break
```

## Bài 14
### Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. Không được dùng mảng.

```python
n = int(input())
while True:
    if n == -1:
        break    
    else:
        max_val = n
        min_val = n
        n = int(input())
        if n > max_val:
            max_val = n
        if n < min_val:
            min_val = n
            print(max_val)
            print(min_val)
```

## Bài 16
### Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ

```python
n = int(input())
even = 0
odd = 0
while n>0:
    if n%2 == 0:
        even += 1
    elif n%2 != 0:
        odd += 1
    n //= 10
print(f"Số chữ số chẵn là: {even}")
print(f"Số chữ số lẻ là: {odd}")
```

## Bài 19
### Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A.

```python
A = int(input())
a,b = 0,1
while a < A:
    a, b = b, b + a
    if b > A:
        print(a)
        break
```

### Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố

```python
n = int(input())
count = 0
while n>0:
    digit = n%10
    if digit in (2, 3, 5, 7):
        count +=1
    n //= 10
print(count)
```
